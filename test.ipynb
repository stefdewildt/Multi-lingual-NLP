```python
import pandas as pd

# 1. Load the dataset
df = pd.read_csv("requested/multitude_v3_clean.csv", encoding="utf-8")

# 2. Filter for Dutch texts (ISO 639-1 code is 'nl')
# Using .str.lower() guards against any capitalization differences
df_dutch = df[df["language"].str.lower() == "nl"].copy()

# Optional: drop the redundant index column if you don't need it
if "Unnamed: 0" in df_dutch.columns:
    df_dutch = df_dutch.drop(columns=["Unnamed: 0"])

# 3. Verify counts (should match expected ~10,344 total: 7,958 train + 2,386 test)
print(f"Dutch samples extracted: {len(df_dutch)}")
print("\nDistribution across splits:")
print(df_dutch["split"].value_counts())

print("\nDistribution across labels:")
print(df_dutch["multi_label"].value_counts())

# 4. Save to a separate CSV file
df_dutch.to_csv("useful/multitude_v3_dutch.csv", index=False, encoding="utf-8")
print("\nSaved to multitude_v3_dutch.csv successfully.")
```
